```python
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import pandas as pd
import numpy.random as random
import sqlite3
from IPython.display import HTML, display, clear_output
import ipywidgets as widgets

# Increase the animation embed limit to prevent frames from being dropped in BVR runs
plt.rcParams['animation.embed_limit'] = 50.0

# ==========================================
# 1. DIGITAL TWIN PHYSICS ENGINE (HARDENED)
# ==========================================
def flight_dynamics(state, t, target_g_limit):
    m_pos = state[0:3]
    m_vel = state[3:6]
    t_pos = state[6:9]
    t_vel = state[9:12]
    a_lat_real = state[12:15]
    m_lift_vector = state[15:18]
    t_a_lat_real = state[18:21]

    R_vec = t_pos - m_pos
    R_mag = np.linalg.norm(R_vec)

    if R_mag < 5.0:
        return np.zeros(21)

    speed_m = np.linalg.norm(m_vel)
    v_hat = np.array([1.0, 0.0, 0.0]) if speed_m == 0 else m_vel / speed_m

    speed_t = np.linalg.norm(t_vel)
    t_v_hat = t_vel / speed_t if speed_t > 0 else np.array([1.0, 0.0, 0.0])

    R_hat = R_vec / max(R_mag, 1e-3)
    V_rel = t_vel - m_vel
    V_close = -np.dot(R_vec, V_rel) / max(R_mag, 1e-3)

    # --- ADVERSARY AI ---
    t_go = R_mag / max(V_close, 1e-3) if V_close > 0 else 999.0
    asset_pos = np.array([0.0, 0.0, 0.0])
    dist_to_base = np.linalg.norm(t_pos - asset_pos)

    if dist_to_base < 3000.0:
        t_cmd_raw = np.array([0.0, 0.0, 0.0])
    elif 15.0 < R_mag < 3500.0 and V_close > 50.0:
        orthogonal_vec = np.cross(R_hat, t_v_hat)
        if np.linalg.norm(orthogonal_vec) == 0:
            orthogonal_vec = np.array([0.0, 1.0, 0.0])
        else:
            orthogonal_vec = orthogonal_vec / np.linalg.norm(orthogonal_vec)
        t_cmd_raw = orthogonal_vec * target_g_limit * 9.81
    elif V_close <= 0.0 and R_mag > 15.0 and t > 5.0:
        asset_dir = asset_pos - t_pos
        asset_hat = asset_dir / (np.linalg.norm(asset_dir) + 1e-9)
        if np.dot(t_v_hat, asset_hat) < 0.99:
            rot_axis = np.cross(t_v_hat, asset_hat)
            if np.linalg.norm(rot_axis) > 0:
                rot_axis = rot_axis / np.linalg.norm(rot_axis)
                t_cmd_raw = np.cross(rot_axis, t_v_hat) * target_g_limit * 9.81
            else:
                t_cmd_raw = np.array([0.0, 0.0, 0.0])
        else:
            t_cmd_raw = np.array([0.0, 0.0, 0.0])
    else:
        t_cmd_raw = np.array([0.0, 0.0, 0.0])

    tau_target = 0.5
    t_a_lat_dot = (t_cmd_raw - t_a_lat_real) / tau_target
    t_vel_dot = t_a_lat_real

    # --- SEEKER & GUIDANCE ---
    cos_theta = np.clip(np.dot(v_hat, R_hat), -1.0, 1.0)
    gimbal_angle_deg = np.degrees(np.arccos(cos_theta))

    if gimbal_angle_deg <= 60.0 and R_mag > 15.0:
        N_const = 4.0
        Omega = np.cross(R_vec, V_rel) / max(R_mag**2, 1e-3)
        a_cmd_raw = N_const * V_close * np.cross(Omega, v_hat)
        if t < 15.0 and R_mag > 40000.0:
            a_cmd_raw[2] += 4.0 * 9.81
        a_cmd_mag = np.linalg.norm(a_cmd_raw)
        if a_cmd_mag > 35.0 * 9.81:
            a_cmd_raw = (a_cmd_raw / a_cmd_mag) * (35.0 * 9.81)
    else:
        a_cmd_raw = np.array([0.0, 0.0, 0.0])

    m_lift_hat = m_lift_vector / (np.linalg.norm(m_lift_vector) + 1e-9)
    a_cmd_dir = a_cmd_raw / (np.linalg.norm(a_cmd_raw) + 1e-9) if np.linalg.norm(a_cmd_raw) > 0 else m_lift_hat
    rot_diff = a_cmd_dir - m_lift_hat
    rot_mag = np.linalg.norm(rot_diff)

    max_rot_rate = np.radians(200.0)
    lift_vector_dot = (rot_diff / max(rot_mag, 1e-9)) * max_rot_rate if rot_mag > max_rot_rate else rot_diff * 5.0

    achievable_a_cmd = np.dot(a_cmd_raw, m_lift_hat) * m_lift_hat
    a_lat_dot = (achievable_a_cmd - a_lat_real) / 0.2

    # --- MASS DEPLETION & KINEMATICS ---
    m_current = 161.5 - (4.5 * t) if t < 7.0 else 130.0
    alt = max(m_pos[2], 0.1)
    T = 288.15 - 0.0065 * alt if alt < 11000 else 216.65
    p = 101325 * (T / 288.15)**5.2561 if alt < 11000 else 22632 * np.exp(-9.81 * (alt - 11000) / (287.05 * 216.65))
    rho = p / (287.05 * T)

    a_sound = np.sqrt(1.4 * 287.05 * T)
    mach = speed_m / a_sound if a_sound > 0 else 0
    Cd = 0.25 if mach < 0.8 else (0.25 + 0.5 * ((mach - 0.8) / 0.4) if mach < 1.2 else max(0.3, 0.75 - 0.3 * ((mach - 1.2) / 2.0)))

    thrust_mag = 45000.0 if t < 2.0 else (12000.0 if t < 7.0 else 0.0)
    F_drag = -(0.5 * rho * (speed_m**2) * Cd * (np.pi * (0.178 / 2)**2)) * v_hat
    F_thrust = thrust_mag * v_hat

    omega_earth = np.array([0.0, 7.2921e-5 * np.cos(np.radians(14.5)), 7.2921e-5 * np.sin(np.radians(14.5))])
    a_coriolis = -2.0 * np.cross(omega_earth, m_vel)

    m_accel = (F_thrust + F_drag) / m_current - np.array([0.0, 0.0, 9.81]) + a_lat_real + a_coriolis
    return np.concatenate((m_vel, m_accel, t_vel, t_vel_dot, a_lat_dot, lift_vector_dot, t_a_lat_dot))

# ==========================================
# 2. ENCAPSULATED SIMULATION PIPELINE
# ==========================================
def run_simulation(threat_type, target_speed):
    print(f"--- INITIALIZING BVR COMBAT RUN: {threat_type} ---")

    t_span = np.linspace(0, 120.0, 1200)
    dt = t_span[1] - t_span[0]
    asset_pos = np.array([0.0, 0.0, 0.0])
    koz_radius = 3000.0

    if threat_type == "SUBSONIC_DRONE":
        target_alt, target_g_limit = random.uniform(3000.0, 6000.0), random.uniform(2.0, 4.0)
    elif threat_type == "FIGHTER_JET":
        target_alt, target_g_limit = random.uniform(8000.0, 15000.0), random.uniform(9.0, 12.0)
    elif threat_type == "SUPERSONIC_CRUISE_MISSILE":
        target_alt, target_g_limit = random.uniform(50.0, 500.0), random.uniform(4.0, 6.0)

    print(f"Cruising Speed: {target_speed:.1f} m/s | Altitude: {target_alt:.0f} m")

    pos_m, vel_m = np.array([0.0, 0.0, 10000.0]), np.array([400.0, 0.0, 0.0])
    pos_t = np.array([random.uniform(80000, 100000), random.uniform(-5000, 5000), target_alt])
    vel_t = ((asset_pos - pos_t) / (np.linalg.norm(asset_pos - pos_t) + 1e-9)) * target_speed

    initial_state = np.concatenate((pos_m, vel_m, pos_t, vel_t, np.zeros(3), np.array([0.0, 0.0, 1.0]), np.zeros(3)))
    telemetry = odeint(flight_dynamics, initial_state, t_span, args=(target_g_limit,), mxstep=5000)

    columns = ['m_x', 'm_y', 'm_z', 'm_vx', 'm_vy', 'm_vz', 't_x', 't_y', 't_z', 't_vx', 't_vy', 't_vz', 'a_x', 'a_y', 'a_z', 'l_x', 'l_y', 'l_z', 'ta_x', 'ta_y', 'ta_z']
    df = pd.DataFrame(telemetry, columns=columns)

    df = df.loc[(df[['m_vx', 'm_vy', 'm_vz']] != 0).any(axis=1)].copy()
    df.reset_index(drop=True, inplace=True)

    df['dist'] = np.sqrt((df['t_x'] - df['m_x'])**2 + (df['t_y'] - df['m_y'])**2 + (df['t_z'] - df['m_z'])**2)
    df['dist_to_asset'] = np.sqrt(df['t_x']**2 + df['t_y']**2 + df['t_z']**2)

    hit_indices = df.index[df['dist'] <= 15.0].tolist()
    fuze_detonated = False

    if hit_indices:
        cpa_idx = hit_indices[0]
        min_miss_distance = df['dist'].iloc[cpa_idx]
        p_kill = 1.0 - (min_miss_distance / 15.0)**2
        kill_roll = random.uniform(0.0, 1.0)
        fuze_detonated = True

        if kill_roll <= p_kill:
            mission_status = f"SUCCESS: TARGET DESTROYED (Pk: {p_kill*100:.1f}%)"
            df = df.iloc[:cpa_idx + 1].copy()
        else:
            base_impact_idx = df['dist_to_asset'].idxmin()
            df = df.iloc[:base_impact_idx + 1].copy()
            koz_status = "KOZ BREACHED" if df['dist_to_asset'].min() <= koz_radius else "NO BREACH"
            mission_status = f"FAILURE: TARGET SURVIVED BLAST (Pk: {p_kill*100:.1f}%) -> {koz_status}"
    else:
        cpa_idx = df['dist'].idxmin()
        base_impact_idx = df['dist_to_asset'].idxmin()
        df = df.iloc[:base_impact_idx + 1].copy()
        mission_status = "FAILURE: KOZ BREACHED" if df['dist_to_asset'].min() <= koz_radius else "FAILURE: KINEMATIC DEFEAT"

    F_mat, H_mat, Q_mat = np.eye(2) + np.array([[0, dt], [0, 0]]), np.array([[1, 0]]), np.array([[1, 0], [0, 3]])
    x_est, P_mat = np.array([[df['t_x'].iloc[0]], [df['t_vx'].iloc[0]]]), np.array([[100, 0], [0, 100]])
    filtered_t_x, sensor_mode = [], []
    pitbull_range = 16000.0

    for idx, row in df.iterrows():
        t_vel_vec = np.array([row['t_vx'], row['t_vy'], row['t_vz']])
        los_vec = np.array([row['t_x'] - row['m_x'], row['t_y'] - row['m_y'], row['t_z'] - row['m_z']])
        los_norm = np.linalg.norm(los_vec)
        los_hat = los_vec / los_norm if los_norm > 1e-6 else np.array([1.0, 0.0, 0.0])
        radial_vel = abs(np.dot(t_vel_vec, los_hat))

        if row['dist'] > pitbull_range:
            variance, mode = 150.0, 'DATALINK_MIDCOURSE'
        elif radial_vel < 35.0:
            variance, mode = 10000.0, 'DOPPLER_NOTCH_LOSS'
        else:
            variance, mode = 20.0, 'ACTIVE_RADAR_PITBULL'

        sensor_mode.append(mode)
        noisy_z = row['t_x'] + random.normal(0, variance)

        x_pred = np.dot(F_mat, x_est)
        P_pred = np.dot(F_mat, np.dot(P_mat, F_mat.T)) + Q_mat
        S = np.dot(H_mat, np.dot(P_pred, H_mat.T)) + np.array([[variance**2]])
        K = np.dot(P_pred, np.dot(H_mat.T, np.linalg.inv(S)))
        x_est = x_pred + np.dot(K, (noisy_z - np.dot(H_mat, x_pred)))
        P_mat = np.dot((np.eye(2) - np.dot(K, H_mat)), P_pred)
        filtered_t_x.append(x_est[0, 0])

    df['sensor_mode'], df['mission_status'] = sensor_mode, mission_status

    fig_anim = plt.figure(figsize=(10, 7))
    ax_anim = fig_anim.add_subplot(111, projection='3d')
    line_m, = ax_anim.plot([], [], [], color='purple', linewidth=2, label='AIM-120')
    line_t, = ax_anim.plot([], [], [], color='orange', linewidth=2, label=threat_type)
    pt_m, pt_t = ax_anim.plot([], [], [], color='purple', marker='o')[0], ax_anim.plot([], [], [], color='orange', marker='o')[0]
    ax_anim.plot([0], [0], [0], color='red', marker='^', markersize=10, label='Defended Asset')
    text_overlay = ax_anim.text2D(0.0, 1.0, '', transform=ax_anim.transAxes, verticalalignment='top', bbox=dict(facecolor='white', alpha=0.8, edgecolor='none'))

    radar_cone = [ax_anim.plot([], [], [], color='red', alpha=0.3, linewidth=1.5)[0] for _ in range(8)]

    detonation_sphere = None
    if fuze_detonated:
        u, v = np.mgrid[0:2*np.pi:20j, 0:np.pi:10j]
        x_sphere = 15.0 * np.cos(u) * np.sin(v) + df['m_x'].iloc[cpa_idx]
        y_sphere = 15.0 * np.sin(u) * np.sin(v) + df['m_y'].iloc[cpa_idx]
        z_sphere = 15.0 * np.cos(v) + df['m_z'].iloc[cpa_idx]
        detonation_sphere = ax_anim.plot_surface(x_sphere, y_sphere, z_sphere, color='red', alpha=0.3, edgecolor='none')
        detonation_sphere.set_visible(False)

    ax_anim.set_xlim([min(df['m_x'].min(), 0)-500, max(df['t_x'].max(), 0)+500])
    ax_anim.set_ylim([min(df['m_y'].min(), 0)-500, max(df['t_y'].max(), 0)+500])
    ax_anim.set_zlim([0, max(df['m_z'].max(), df['t_z'].max())+500])
    ax_anim.legend(loc='upper left', bbox_to_anchor=(1.05, 1.0))
    plt.subplots_adjust(right=0.75)

    frame_list = list(range(0, len(df), 10))
    if len(frame_list) == 0: frame_list = [0]
    elif frame_list[-1] != len(df) - 1: frame_list.append(len(df) - 1)

    def update(num):
        plot_num_m = cpa_idx if (fuze_detonated and num >= cpa_idx) else num
        line_m.set_data(df['m_x'][:plot_num_m], df['m_y'][:plot_num_m])
        line_m.set_3d_properties(df['m_z'][:plot_num_m])

        line_t.set_data(df['t_x'][:num], df['t_y'][:num])
        line_t.set_3d_properties(df['t_z'][:num])

        mx, my, mz = df['m_x'].iloc[plot_num_m], df['m_y'].iloc[plot_num_m], df['m_z'].iloc[plot_num_m]
        pt_m.set_data([mx], [my])
        pt_m.set_3d_properties([mz])

        if fuze_detonated and num >= cpa_idx:
            pt_m.set_visible(False)
        else:
            pt_m.set_visible(True)

        pt_t.set_data([df['t_x'].iloc[num]], [df['t_y'].iloc[num]])
        pt_t.set_3d_properties([df['t_z'].iloc[num]])

        if num >= cpa_idx and fuze_detonated:
            display_status = mission_status
            mode = "OFFLINE"
        else:
            display_status = "IN FLIGHT (Tracking)"
            mode = df['sensor_mode'].iloc[num]

        # --- DYNAMIC ACTIVE RADAR CONE RENDERER (GIMBAL TRACKING) ---
        if mode == 'ACTIVE_RADAR_PITBULL':
            # Calculate the true Line-of-Sight (LOS) vector directly to the target
            los_vec = np.array([df['t_x'].iloc[num] - mx, df['t_y'].iloc[num] - my, df['t_z'].iloc[num] - mz])
            dist_to_target = np.linalg.norm(los_vec)
            los_hat = los_vec / dist_to_target if dist_to_target > 1e-6 else np.array([1.0, 0.0, 0.0])

            # Anchor the cone to the gimballed LOS vector instead of the missile's velocity vector
            u1 = np.cross(los_hat, np.array([1, 0, 0])) if np.abs(los_hat[0]) < 0.9 else np.cross(los_hat, np.array([0, 1, 0]))
            u1 = u1 / np.linalg.norm(u1)
            u2 = np.cross(los_hat, u1)

            # Tight 4-degree FOV perfectly brackets the target like a hard-lock
            fov_rad = np.radians(4.0)
            cone_len = dist_to_target

            for i in range(8):
                angle = i * (2 * np.pi / 8)
                dir_vec = los_hat * np.cos(fov_rad) + (u1 * np.cos(angle) + u2 * np.sin(angle)) * np.sin(fov_rad)
                end_pt = np.array([mx, my, mz]) + dir_vec * cone_len

                radar_cone[i].set_data([mx, end_pt[0]], [my, end_pt[1]])
                radar_cone[i].set_3d_properties([mz, end_pt[2]])
                radar_cone[i].set_visible(True)
        else:
            for line in radar_cone:
                line.set_data(np.array([]), np.array([]))
                line.set_3d_properties(np.array([]))
                line.set_visible(False)

        if detonation_sphere is not None and num >= cpa_idx:
            detonation_sphere.set_visible(True)

        text_overlay.set_text(f"Status: {display_status}\n"
                              f"Dist to Target: {df['dist'].iloc[num]:.1f}m\n"
                              f"Dist to Base: {df['dist_to_asset'].iloc[num]:.1f}m\n"
                              f"Sensor: {mode}")
        return [line_m, line_t, pt_m, pt_t, text_overlay] + radar_cone

    ani = animation.FuncAnimation(fig_anim, update, frames=frame_list, interval=35)
    plt.close(fig_anim)
    display(HTML(ani.to_jshtml()))

# ==========================================
# 3. IPYWIDGETS INTERACTIVE DASHBOARD
# ==========================================
threat_dropdown = widgets.Dropdown(
    options=["SUBSONIC_DRONE", "FIGHTER_JET", "SUPERSONIC_CRUISE_MISSILE"],
    value="FIGHTER_JET",
    description="Target Profile:",
    style={'description_width': 'initial'}
)

speed_slider = widgets.FloatSlider(
    value=300.0, min=250.0, max=400.0, step=5.0,
    description="Target Speed (m/s):", style={'description_width': 'initial'},
    layout=widgets.Layout(width='400px')
)

run_button = widgets.Button(description="Run Intercept Simulation", button_style='danger', icon='rocket')
output_area = widgets.Output()

def update_speed_bounds(*args):
    selection = threat_dropdown.value
    speed_slider.min = 0.0
    speed_slider.max = 2000.0
    if selection == "SUBSONIC_DRONE":
        speed_slider.min, speed_slider.max, speed_slider.value = 100.0, 150.0, 125.0
    elif selection == "FIGHTER_JET":
        speed_slider.min, speed_slider.max, speed_slider.value = 250.0, 400.0, 300.0
    elif selection == "SUPERSONIC_CRUISE_MISSILE":
        speed_slider.min, speed_slider.max, speed_slider.value = 700.0, 900.0, 800.0

threat_dropdown.observe(update_speed_bounds, 'value')

def on_run_clicked(b):
    with output_area:
        clear_output(wait=True)
        run_simulation(threat_dropdown.value, speed_slider.value)

run_button.on_click(on_run_clicked)
display(widgets.VBox([threat_dropdown, speed_slider, run_button, output_area]))
```

```python
import numpy as np
from scipy.integrate import odeint
import pandas as pd
import numpy.random as random
import sqlite3
from IPython.display import clear_output
import time

# ==========================================
# 1. THE BENCHMARK PHYSICS ENGINE
# ==========================================
def flight_dynamics(state, t, target_g_limit):
    m_pos = state[0:3]
    m_vel = state[3:6]
    t_pos = state[6:9]
    t_vel = state[9:12]
    a_lat_real = state[12:15]
    m_lift_vector = state[15:18]
    t_a_lat_real = state[18:21]

    R_vec = t_pos - m_pos
    R_mag = np.linalg.norm(R_vec)

    if R_mag < 5.0:
        return np.zeros(21)

    speed_m = np.linalg.norm(m_vel)
    v_hat = np.array([1.0, 0.0, 0.0]) if speed_m == 0 else m_vel / speed_m

    speed_t = np.linalg.norm(t_vel)
    t_v_hat = t_vel / speed_t if speed_t > 0 else np.array([1.0, 0.0, 0.0])

    R_hat = R_vec / R_mag
    V_rel = t_vel - m_vel
    V_close = -np.dot(R_vec, V_rel) / R_mag

    t_go = R_mag / V_close if V_close > 0 else 999.0
    asset_pos = np.array([0.0, 0.0, 0.0])

    if 15.0 < R_mag < 3500.0 and V_close > 50.0:
        orthogonal_vec = np.cross(R_hat, t_v_hat)
        if np.linalg.norm(orthogonal_vec) == 0:
            orthogonal_vec = np.array([0.0, 1.0, 0.0])
        else:
            orthogonal_vec = orthogonal_vec / np.linalg.norm(orthogonal_vec)
        t_cmd_raw = orthogonal_vec * target_g_limit * 9.81

    elif V_close <= 0.0 and R_mag > 15.0 and t > 5.0:
        asset_dir = asset_pos - t_pos
        asset_hat = asset_dir / (np.linalg.norm(asset_dir) + 1e-9)

        if np.dot(t_v_hat, asset_hat) < 0.99:
            rot_axis = np.cross(t_v_hat, asset_hat)
            if np.linalg.norm(rot_axis) > 0:
                rot_axis = rot_axis / np.linalg.norm(rot_axis)
                t_cmd_raw = np.cross(rot_axis, t_v_hat) * target_g_limit * 9.81
            else:
                t_cmd_raw = np.array([0.0, 0.0, 0.0])
        else:
            t_cmd_raw = np.array([0.0, 0.0, 0.0])
    else:
        t_cmd_raw = np.array([0.0, 0.0, 0.0])

    tau_target = 0.5
    t_a_lat_dot = (t_cmd_raw - t_a_lat_real) / tau_target
    t_vel_dot = t_a_lat_real

    cos_theta = np.clip(np.dot(v_hat, R_hat), -1.0, 1.0)
    gimbal_angle_deg = np.degrees(np.arccos(cos_theta))

    if gimbal_angle_deg <= 60.0 and R_mag > 15.0:
        N_const = 4.0
        Omega = np.cross(R_vec, V_rel) / (R_mag**2)
        a_cmd_raw = N_const * V_close * np.cross(Omega, v_hat)

        if t < 15.0 and R_mag > 40000.0:
            a_cmd_raw[2] += 4.0 * 9.81

        a_cmd_mag = np.linalg.norm(a_cmd_raw)
        if a_cmd_mag > 35.0 * 9.81:
            a_cmd_raw = (a_cmd_raw / a_cmd_mag) * (35.0 * 9.81)
    else:
        a_cmd_raw = np.array([0.0, 0.0, 0.0])

    m_lift_hat = m_lift_vector / (np.linalg.norm(m_lift_vector) + 1e-9)
    a_cmd_dir = a_cmd_raw / (np.linalg.norm(a_cmd_raw) + 1e-9) if np.linalg.norm(a_cmd_raw) > 0 else m_lift_hat
    rot_diff = a_cmd_dir - m_lift_hat
    rot_mag = np.linalg.norm(rot_diff)

    max_rot_rate = np.radians(200.0)
    lift_vector_dot = (rot_diff / (rot_mag + 1e-9)) * max_rot_rate if rot_mag > max_rot_rate else rot_diff * 5.0

    achievable_a_cmd = np.dot(a_cmd_raw, m_lift_hat) * m_lift_hat

    tau = 0.2
    a_lat_dot = (achievable_a_cmd - a_lat_real) / tau

    m_current = 161.5 - (4.5 * t) if t < 7.0 else 130.0

    alt = max(m_pos[2], 0.1)
    T = 288.15 - 0.0065 * alt if alt < 11000 else 216.65
    p = 101325 * (T / 288.15)**5.2561 if alt < 11000 else 22632 * np.exp(-9.81 * (alt - 11000) / (287.05 * 216.65))
    rho = p / (287.05 * T)

    a_sound = np.sqrt(1.4 * 287.05 * T)
    mach = speed_m / a_sound if a_sound > 0 else 0
    Cd = 0.25 if mach < 0.8 else (0.25 + 0.5 * ((mach - 0.8) / 0.4) if mach < 1.2 else max(0.3, 0.75 - 0.3 * ((mach - 1.2) / 2.0)))

    thrust_mag = 45000.0 if t < 2.0 else (12000.0 if t < 7.0 else 0.0)
    A = np.pi * (0.178 / 2)**2
    F_drag = -(0.5 * rho * (speed_m**2) * Cd * A) * v_hat
    F_thrust = thrust_mag * v_hat

    omega_earth = np.array([0.0, 7.2921e-5 * np.cos(np.radians(14.5)), 7.2921e-5 * np.sin(np.radians(14.5))])
    a_coriolis = -2.0 * np.cross(omega_earth, m_vel)
    g = np.array([0.0, 0.0, 9.81])
    m_accel = (F_thrust + F_drag) / m_current - g + a_lat_real + a_coriolis

    return np.concatenate((m_vel, m_accel, t_vel, t_vel_dot, a_lat_dot, lift_vector_dot, t_a_lat_dot))


# ==========================================
# 2. THE MONTE CARLO BATCH PROCESSOR
# ==========================================
def run_monte_carlo(num_simulations=500):
    t_span = np.linspace(0, 120.0, 1200)
    koz_radius = 3000.0
    asset_pos = np.array([0.0, 0.0, 0.0])

    batch_results = []
    start_time = time.time()

    for i in range(num_simulations):
        # 1. Randomize Engagement Parameters
        threat_types = ["SUBSONIC_DRONE", "FIGHTER_JET", "SUPERSONIC_CRUISE_MISSILE"]
        threat = random.choice(threat_types)

        if threat == "SUBSONIC_DRONE":
            target_speed = random.uniform(100.0, 150.0)
            target_alt = random.uniform(3000.0, 6000.0)
            target_g_limit = random.uniform(2.0, 4.0)
        elif threat == "FIGHTER_JET":
            target_speed = random.uniform(250.0, 400.0)
            target_alt = random.uniform(8000.0, 15000.0)
            target_g_limit = random.uniform(9.0, 12.0)
        else:
            target_speed = random.uniform(700.0, 900.0)
            target_alt = random.uniform(50.0, 500.0)
            target_g_limit = random.uniform(4.0, 6.0)

        # Randomize launch range between 20km (WVR) and 120km (Extreme BVR)
        launch_range = random.uniform(20000.0, 120000.0)

        pos_m = np.array([0.0, 0.0, 10000.0])
        vel_m = np.array([400.0, 0.0, 0.0])

        pos_t = np.array([launch_range, random.uniform(-5000, 5000), target_alt])
        target_dir = asset_pos - pos_t
        vel_t = (target_dir / (np.linalg.norm(target_dir) + 1e-9)) * target_speed

        initial_state = np.concatenate((pos_m, vel_m, pos_t, vel_t, np.zeros(3), np.array([0.0, 0.0, 1.0]), np.zeros(3)))

        # 2. Run the Solver
        telemetry = odeint(flight_dynamics, initial_state, t_span, args=(target_g_limit,), mxstep=5000)

        # 3. Fast Array Operations (No Pandas overhead during the loop)
        m_x, m_y, m_z = telemetry[:, 0], telemetry[:, 1], telemetry[:, 2]
        t_x, t_y, t_z = telemetry[:, 6], telemetry[:, 7], telemetry[:, 8]

        dist_array = np.sqrt((t_x - m_x)**2 + (t_y - m_y)**2 + (t_z - m_z)**2)
        dist_to_asset_array = np.sqrt(t_x**2 + t_y**2 + t_z**2)

        min_miss_dist = np.min(dist_array)
        min_asset_dist = np.min(dist_to_asset_array)

        # 4. Evaluate Lethality & Mission Success
        outcome = 0 # 0 = Failure, 1 = Success
        p_kill = 0.0

        if min_miss_dist <= 15.0:
            p_kill = 1.0 - (min_miss_dist / 15.0)**2
            if random.uniform(0.0, 1.0) <= p_kill:
                outcome = 1

        koz_breach = 1 if min_asset_dist <= koz_radius and outcome == 0 else 0

        # 5. Log the Data
        batch_results.append({
            'sim_id': i + 1,
            'threat_profile': threat,
            'target_speed_mps': round(target_speed, 1),
            'target_alt_m': round(target_alt, 1),
            'evasive_g_limit': round(target_g_limit, 1),
            'launch_range_m': round(launch_range, 1),
            'min_miss_dist_m': round(min_miss_dist, 2),
            'p_kill': round(p_kill, 3),
            'mission_success': outcome,
            'koz_breached': koz_breach
        })

        # Update Progress Bar
        if (i + 1) % 25 == 0:
            clear_output(wait=True)
            elapsed = time.time() - start_time
            print(f"Running Monte Carlo... {i + 1}/{num_simulations} completed. ({elapsed:.1f}s elapsed)")

    # 6. Export to DataFrame and SQL
    df_batch = pd.DataFrame(batch_results)

    conn = sqlite3.connect('batch_results.db')
    df_batch.to_sql('monte_carlo_data', conn, if_exists='replace', index=False)
    conn.close()

    clear_output(wait=True)
    print(f"--- BATCH COMPLETE ---")
    print(f"Total Simulations: {num_simulations}")
    print(f"Overall Success Rate: {(df_batch['mission_success'].mean() * 100):.1f}%")
    print("Dataset exported to 'batch_results.db' and stored in 'df_batch' variable.")

    return df_batch

# Run the batch processor (Adjust num_simulations as desired)
df_batch = run_monte_carlo(num_simulations=500)
display(df_batch.head(500))
```
